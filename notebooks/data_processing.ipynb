## Demographic Dependency Index Processing

```python
# Import the CSV file containing population data
# Rename the columns of the dataset
# Skip the first row from the file (original header)
df_pop = pd.read_csv('Data/Population/exportPivot_POP105A_2.csv', delimiter =',', names = ['Age', 'Sex', 'Areas', 'County', 'Years', 'Persons', 'Population'], skiprows = 1)

# Display the resulting DataFrame
df_pop
```

```python
# Remove unnecessary columns from the DataFrame
df_pop = df_pop.drop(['Sex', 'Areas', 'Persons'], axis = 1)
```

```python
# Remove the 'Year' prefix from the values in the 'Years' column
df_pop['Years'] = df_pop['Years'].str.lstrip('Year ')

# Display the updated DataFrame
df_pop
```

```python
# Remove leading spaces from the values in the 'County' column
df_pop['County'] = df_pop['County'].str.lstrip(' ')
```

```python
df_pop
```

```python
# Define cateogories
young = ['0- 4 years', '5- 9 years', '10-14 years']
```

```python
working = [
    '15-19 years', '20-24 years', '25-29 years',
    '30-34 years', '35-39 years', '40-44 years',
    '45-49 years', '50-54 years', '55-59 years',
    '60-64 years']
```

```python
old = [
    '65-69 years', '70-74 years', '75-79 years',
    '80-84 years', '85 years and over']
```

```python
# Create demographic category
def classify_age(age):
    if age in young:
        return 'young'
    elif age in working:
        return 'working'
    elif age in old:
        return 'old'
    else:
        return None

# Apply the classification function to the 'Age' column and create a new column
df_pop['Category'] = df_pop['Age'].apply(classify_age)
```

```python
# Create a pivot table with population values grouped by county, year, and demographic category
df_pivot_pop = df_pop.pivot_table(values = 'Population', index = ['County', 'Years'], columns = 'Category', aggfunc = 'sum').reset_index()
```

```python
# Dependency index calculation 
df_pivot_pop['Dependency index'] = ((df_pivot_pop['young'] + df_pivot_pop ['old']) / df_pivot_pop['working']) * 100
```

```python
df_pivot_pop
```

```python
# Export the dependency index DataFrame to an Excel file without including the index
df_pivot_pop.to_excel('dependency_index_county_2013_2024.xlsx', columns = ['County', 'Years', 'Dependency index'], index = False)
```

## Salary processing

```python
# Import the CSV file containing salary data
# Rename the columns of the dataset
# Skip the first row from the file (original header)
df_salaries = pd.read_csv('Data/Salary/exportPivot_FOM106E.csv', names = ['NACE', 'Sex', 'County', 'Years', 'RON', 'Salary'], skiprows = 1)

# Display the resulting DataFram
df_salaries
```

```python
# Remove unnecessary columns from the DataFrame
df_salaries = df_salaries.drop(['NACE', 'Sex', 'RON'], axis = 1)

# Display the cleaned DataFrame
df_salaries
```

```python
# Remove the 'Year' prefix from the values in the 'Years' column
df_salaries['Years'] = df_salaries['Years'].str.lstrip('Year ')
```

```python
# Remove leading spaces from the values in the 'County' column
df_salaries['County'] = df_salaries['County'].str.lstrip(' ')

# Display the updated DataFrame
df_salaries
```

```python
# Import the Excel file containing inflation data
df_inflation = pd.read_excel('Data/Salary/inflation_ins.xlsx')

# Display the resulting DataFrame
df_inflation
```

```python
# Sort the inflation data by year and reset the index
df_inflation = df_inflation.sort_values("Years").reset_index(drop=True)

# Create a cumulative price index with 2013 as the base year
# The value 100 means that prices in 2013 are treated as the reference level
df_inflation["Price_Index_2013_100"] = 100.0

# Calculate the cumulative price index for each year
# Each year's index is obtained by multiplying the previous year's index
# by the Consumer Price Index (IPC) of the current year
for i in range(1, len(df_inflation)):
    df_inflation.loc[i, "Price_Index_2013_100"] = (
        df_inflation.loc[i-1, "Price_Index_2013_100"] *
        df_inflation.loc[i, "IPC"] / 100
    )

# Extract the cumulative price index value for 2024
# This will be used as the new reference value
base_2024 = df_inflation.loc[
    df_inflation["Years"] == 2024,
    "Price_Index_2013_100"
].iloc[0]

# Rebase the price index so that 2024 becomes the reference year
# A value below 100 means that prices were lower than in 2024
df_inflation["Price_Index_2024_100"] = (
    df_inflation["Price_Index_2013_100"] /
    base_2024
) * 100

# Display the DataFrame
df_inflation
```

```python
# Convert the values in the 'Years' column to string type
df_inflation['Years'] = df_inflation['Years'].astype(str)
```

```python
# Merge the salary and inflation DataFrames based on the 'Years' column
df_real_salary = pd.merge(df_salaries, df_inflation, on = "Years", how="inner")

# Display the merged DataFrame
df_real_salary
```

```python
# Calculate the real salary adjusted for inflation using 2024 as the reference year
df_real_salary['Real salary'] = (df_real_salary['Salary'] / df_real_salary['Price_Index_2024_100']) * 100 
```

```python
# Display the first 12 rows of the DataFrame
df_real_salary.head(12)
```

```python
# Export the real salary DataFrame to an Excel file without including the index
# Save salary values without decimal places
df_real_salary.to_excel('real_salary_county_2013_2024.xlsx', index = False, columns = ['County', 'Years', 'Real salary'], float_format = '%.0f')
```
