```python
# importing libraries
import pandas as pd 
import numpy as np
```

```python
# Import dependency index dataset
dependency_index_county_2013_2024 = pd.read_excel('dependency_index_county_2013_2024.xlsx')

# Display dataset
dependency_index_county_2013_2024
```

```python
# Remove leading spaces from county names
dependency_index_county_2013_2024['County'] = dependency_index_county_2013_2024['County'].str.lstrip(' ')
```

```python
# Import enterprises per capita dataset
enterprises_capita_county_2013_2024 = pd.read_excel('enterprises_capita_county_2013_2024.xlsx')

# Display dataset
enterprises_capita_county_2013_2024
```

```python
# Remove leading spaces from county names
enterprises_capita_county_2013_2024['County'] = enterprises_capita_county_2013_2024['County'].str.lstrip(' ')
```

```python
# Import FDI per capita dataset
fdi_capita_county_2013_2024 = pd.read_excel('fdi_capita_county_2013_2024.xlsx')

# Display dataset
fdi_capita_county_2013_2024
```

```python
# Remove leading spaces from county names
fdi_capita_county_2013_2024['County'] = fdi_capita_county_2013_2024['County'].str.lstrip(' ')
```

```python
# Import occupancy rate dataset
occupancy_rate_county_2013_2024 = pd.read_excel('occupancy_rate_county_2013_2024.xlsx')

# Display dataset
occupancy_rate_county_2013_2024
```

```python
# Remove leading spaces from county names
occupancy_rate_county_2013_2024['County'] = occupancy_rate_county_2013_2024['County'].str.lstrip(' ')
```

```python
# Import real salary dataset
real_salary_county_2013_2024 = pd.read_excel('real_salary_county_2013_2024.xlsx')

# Display dataset
real_salary_county_2013_2024
```

```python
# Remove leading spaces from county names
real_salary_county_2013_2024['County'] = real_salary_county_2013_2024['County'].str.lstrip(' ')
```

```python
# Import temporary immigrants dataset
temporary_immigrants_county_2013_2024 = pd.read_excel('temporary_immigrants_county_2013_2024.xlsx')

# Display dataset
temporary_immigrants_county_2013_2024
```

```python
# Remove leading spaces from county names
temporary_immigrants_county_2013_2024['County'] = temporary_immigrants_county_2013_2024['County'].str.lstrip(' ')
```

```python
# Merge dependency index and enterprises datasets
df_panel = pd.merge(
    dependency_index_county_2013_2024,
    enterprises_capita_county_2013_2024,
    how = 'inner',
    on = ['County', 'Years']
)

# Merge FDI dataset
df_panel = pd.merge(
    df_panel,
    fdi_capita_county_2013_2024,
    how = 'inner',
    on = ['County', 'Years']
)

# Merge occupancy rate dataset
df_panel = pd.merge(
    df_panel,
    occupancy_rate_county_2013_2024,
    how = 'inner',
    on = ['County', 'Years']
)

# Merge real salary dataset
df_panel = pd.merge(
    df_panel,
    real_salary_county_2013_2024,
    how = 'inner',
    on = ['County', 'Years']
)

# Merge temporary immigrants dataset
df_panel = pd.merge(
    df_panel,
    temporary_immigrants_county_2013_2024,
    how = 'inner',
    on = ['County', 'Years']
)

# Display final panel dataset
df_panel
```

```python
# Export panel dataset to Excel
df_panel.to_excel('panel_data.xlsx', index = False)
```
